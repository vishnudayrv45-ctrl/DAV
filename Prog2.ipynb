```python
#loading dataset and printing information from it
import pandas as pd

df = pd.read_csv(r"C:\Users\SHREYA C RAMANI\Downloads\archive\student_performance.csv")

print("Dataset loaded successfully!")

print("In First 5 rows:")
print(df.head())

print("In Last 5 rows:")
print(df.tail())

print("In Shape of the dataset:")
print(df.shape)

print("In Column names:")
print(df.columns)

print("In Data types:")
print(df.dtypes)

print("In Descriptive Statistics:")
print(df.describe(include="all"))

print("In Missing values in each column:")
print(df.isnull().sum())

print("In Number of duplicate rows:")
print(df.duplicated().sum())

print("In Number of duplicate student ID:")
print(df["student_id"].duplicated().sum())

print("In Number of students in each grade:")
print(df["grade"].value_counts())

grade_info = pd.DataFrame({
    "grade": ["A", "B", "C", "D", "F"],
    "grade_description": ["Excellent", "Good", "Average", "Below Average", "Fail"]
})

print("In Grade Information DataFrame:")
print(grade_info)

merged = pd.merge(df, grade_info, on="grade", how="left")

print("In Merged Dataframe:")
print(merged.head())

grade_details = pd.DataFrame({
    "performance": ["Excellent", "Good", "Average", "Below Average", "Fail"],
    "performance_level": ["Very High", "High", "Medium", "Low", "Very low"]
}, index=["A", "B", "C", "D", "F"])

print(grade_details.head())

grade_counts = df["grade"].value_counts().to_frame("student_count")
print(grade_counts.head())

joined = grade_counts.join(grade_details)
print(joined)

df_1 = df.iloc[:100]
df_2 = df.iloc[100:200]

combined = pd.concat([df_1, df_2])
print(combined.head())
print(combined.shape)
```
