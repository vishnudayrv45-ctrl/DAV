```python
#dataframes and accessing index positions,slicing, using numpy and pandas libraries
import numpy as np
import pandas as pd

# NumPy Operations
arr = np.array([10, 20, 30, 40, 50])
print("Array : ", arr)
print("Mean : ", arr.mean())
print("Reshaped 2D :\n", arr.reshape(5, 1))
print("Sliced (index 1 to 3) : ", arr[1:4])

# Pandas DataFrame Operations
data = {
    "Name": ["Asha", "Ravi", "Mitra", "Karan"],
    "Marks":[88,72,91,65],
    "Branch": ["CSE", "ECE", "CSE", "ISE"]
}

df = pd.DataFrame(data)
print(df)

print("\n First 2 Rows \n", df.iloc[0:2])
print("\n Marks column \n", df["Marks"])
print("\n Rows where Marks > 70 \n", df[df["Marks"] > 70])
```
